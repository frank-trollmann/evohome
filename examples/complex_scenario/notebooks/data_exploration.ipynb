## Data Loading

Below we load the dataset and observe it's general properties.

For easier visualization, the dataset is resampled into a dataset "day_averages" with daily averages.

```python
# Load scenario information. This assumes the scenario hasn't changed since data was recorded.
scenario = create_complex_scenario()
room_names = scenario.get_room_names()
nr_rooms = len(room_names)

# load data and drop unused labels.
complete_data = pd.read_pickle("examples/complex_scenario/data/running.pickle")
complete_data.set_index(pd.DatetimeIndex(complete_data["date"]), inplace=True)
prediction_labels = [f"prediction_{nr}" for nr in range(nr_rooms)]
complete_data.drop(labels = ["prediction_time","adaptation_time"] + prediction_labels, axis = 1, inplace= True)

# create day-averaged
day_averages = complete_data.resample("D").mean()

# print statistics
print(f"The dataframe has {len(complete_data)} samples.")
print(f"The example scenario has {nr_rooms} rooms.")
day_averages
```

## Room Occupation Heatmap

Here we visualize the occupancy of each room as a heatmap over each day. 

Here we visualize the occupancy of rooms based on a heatmap.

```python
nr_cols = 2
fig, subplots = plt.subplots(ncols= nr_cols,nrows= (nr_rooms+1) // nr_cols, figsize = (8,200))

for room_index in range(nr_rooms):
   ax = subplots[room_index//nr_cols][room_index% nr_cols]
   ax.set_title(f"Room {room_names[room_index]}")
   sns.heatmap(data =complete_data.groupby([complete_data.date.dt.strftime('%Y-%m-%d'), complete_data.date.dt.strftime('%H:00')])
      [f"room_{room_index}"].mean()
      .rename_axis(index=['day','hour'])
      .unstack(level=0),
      vmin= 0, 
      vmax = 1,
      cmap=LinearSegmentedColormap.from_list('rg',["w", "g"], N=256),
      ax=ax
   )

fig.tight_layout()
plt.show()
```

## Average Room Occupation
Below we visualize the day-averages of occupation for all rooms in the timeframe.

```python

nr_cols = 2
fig, subplots = plt.subplots(ncols= nr_cols,nrows= (nr_rooms+1) //nr_cols, figsize = (16,200))
for room_index in range(nr_rooms):
    ax = subplots[room_index//nr_cols][room_index% nr_cols]
    ax.set_title(f"Room {room_names[room_index]}")
    ax.plot(day_averages[f"room_{room_index}"],"r-")
    ax.set_ylim([0, 1.01])
    ax.xaxis.set_major_locator(mdates.MonthLocator(interval = 1))
    
    
fig.tight_layout()
fig.autofmt_xdate()
plt.xticks(rotation=90) 
plt.show()
```
